```python

faq_data = [
    {
        "question": "What documents are required for admission?",
        "answer": "Please check the official admission guidelines for the required documents."
    },
    {
        "question": "How can I check my attendance?",
        "answer": "Please log in to your student portal and open the attendance section."
    },
    {
        "question": "How can I pay my fees?",
        "answer": "Please use the official student portal or contact the accounts department."
    },
    {
        "question": "How can I contact student support?",
        "answer": "Please contact your institution's student support team."
    },
    {
        "question": "Where can I find my course details?",
        "answer": "Please log in to your student portal and open the courses section."
    }
]

print("Student information loaded!")
print("Number of FAQs:", len(faq_data))
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Collect the questions from our FAQ list
faq_questions = [
    item["question"] for item in faq_data
]

# Convert the questions into numerical features
vectorizer = TfidfVectorizer()
faq_vectors = vectorizer.fit_transform(faq_questions)

print("The chatbot has learned to compare questions!")
```

```python

def get_answer(user_question):
    # Convert the student's question into numerical features
    question_vector = vectorizer.transform([user_question])

    # Compare it with all FAQ questions
    similarities = cosine_similarity(
        question_vector, faq_vectors
    ).flatten()

    # Find the closest matching question
    best_match_index = similarities.argmax()
    best_score = similarities[best_match_index]

    # Avoid answering if the match is too weak
    if best_score < 0.15:
        return (
            "Sorry, I couldn't find a suitable answer. "
            "Please contact student support."
        )

    return faq_data[best_match_index]["answer"]


print(get_answer("How do I see my attendance?"))
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

# Create the chat interface
title = widgets.HTML(
    "<h2>🎓 AI Student Assistant</h2>"
    "<p>Ask me about admissions, fees, courses, or attendance.</p>"
)

question_box = widgets.Text(
    placeholder="Type your question here...",
    description="Question:",
    layout=widgets.Layout(width="85%")
)

ask_button = widgets.Button(
    description="Ask",
    button_style="primary"
)

chat_output = widgets.Output()

def ask_chatbot(button):
    question = question_box.value.strip()

    with chat_output:
        if not question:
            print("Please type a question first.")
            return

        answer = get_answer(question)

        print("You:", question)
        print("Assistant:", answer)
        print("-" * 50)

    question_box.value = ""

ask_button.on_click(ask_chatbot)

display(title, question_box, ask_button, chat_output)
```
